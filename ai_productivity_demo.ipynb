```python
# Install required libraries
!pip install -q google-genai

import os
from google import genai

# Initialize client (Ensure GEMINI_API_KEY is set in your Colab Secrets)
client = genai.Client(api_key=os.environ.get("GEMINI_API_KEY"))
```

```python
# Few-Shot Prompting for Sentiment Classification
prompt = """
Classify the sentiment of customer reviews as Positive, Neutral, or Negative.

Review: \"The battery life is decent, but the screen scratches way too easily.\" -> Negative
Review: \"Absolute game changer! Saved me hours of manual coding.\" -> Positive
Review: \"It works fine, nothing extraordinary.\" -> Neutral
Review: \"Customer support took 3 days to reply, very frustrating.\" -> 
"""

response = client.models.generate_content(
    model='gemini-2.5-flash',
    contents=prompt,
)
print("--- Few-Shot Classification Output ---")
print(response.text)
```

```python
# Structured JSON Output Generation
json_prompt = """
Extract product details from the text below and return ONLY a valid JSON object with keys: \"product_name\", \"price\", and \"rating\".

Text: \"Acme Ultra Mechanical Keyboard priced at ₹4,999, rated 4.7 stars.\"
"""

json_response = client.models.generate_content(
    model='gemini-2.5-flash',
    contents=json_prompt,
)
print("\n--- Structured JSON Output ---")
print(json_response.text)
```
